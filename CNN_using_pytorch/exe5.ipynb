
# Convolutional Neural Networks (CNNs) in Python with PyTorch

## 1. Introduction

A **Convolutional Neural Network (CNN)** is a type of deep learning model primarily designed to process grid-like data, especially images. Unlike fully connected neural networks, CNNs can automatically learn spatial features such as edges, textures, shapes, and objects.

CNNs use convolutional layers to extract features from input images, pooling layers to reduce spatial dimensions, and fully connected layers to perform classification.

PyTorch provides the `torch.nn` module to build and train CNN architectures.

## 2. Mathematical Background

### 2.1 Input Image Representation

A digital image can be represented as a tensor:

$$
X \in \mathbb{R}^{C \times H \times W}
$$

where:

- $C$: Number of channels (e.g., 1 for grayscale and 3 for RGB).
- $H$: Image height.
- $W$: Image width.

For a batch of images, the input tensor has the shape:

$$
X \in \mathbb{R}^{N \times C \times H \times W}
$$

where $N$ represents the batch size.

### 2.2 Convolution Operation

The convolution operation applies a small matrix called a **filter** or **kernel** to an input image to extract local features.

For a two-dimensional input, the convolution operation can be expressed as:

$$
Y(i,j)=
\sum_{m=0}^{K_h-1}
\sum_{n=0}^{K_w-1}
X(i+m,j+n)K(m,n)+b
$$

where:

- $X$: Input image or feature map.
- $K$: Convolution kernel.
- $K_h,K_w$: Kernel height and width.
- $b$: Bias.
- $Y$: Output feature map.

Different filters learn to detect different patterns, such as edges, corners, and textures.

### 2.3 Output Dimension

The spatial output size of a convolutional layer is calculated by:

$$
H_{\text{out}} =
\left\lfloor
\frac{H+2P-D(K_h-1)-1}{S}+1
\right\rfloor
$$

Similarly,

$$
W_{\text{out}} =
\left\lfloor
\frac{W+2P-D(K_w-1)-1}{S}+1
\right\rfloor
$$

where:

- $P$: Padding.
- $S$: Stride.
- $D$: Dilation.
- $K_h,K_w$: Kernel dimensions.

### 2.4 Activation Function

After convolution, a nonlinear activation function is usually applied. A common choice is ReLU:

$$
\operatorname{ReLU}(x)=\max(0,x)
$$

ReLU introduces nonlinearity, allowing the network to learn complex patterns.

### 2.5 Pooling Layer

Pooling reduces the spatial dimensions of feature maps while retaining important information.

A common operation is **Max Pooling**, which selects the maximum value within a local region:

$$
Y(i,j)=\max_{(m,n)\in R_{i,j}}X(m,n)
$$

For example, a $2\times2$ max-pooling operation with stride 2 reduces the height and width by approximately half.

## 3. Basic CNN Architecture

A typical CNN consists of the following sequence:

1. **Input layer:** Receives image tensors.
2. **Convolutional layer:** Extracts spatial features using learnable filters.
3. **Activation layer:** Introduces nonlinear transformations.
4. **Pooling layer:** Reduces spatial dimensions.
5. **Flattening layer:** Converts feature maps into a one-dimensional vector.
6. **Fully connected layer:** Combines extracted features for classification.
7. **Output layer:** Produces class logits.

## 4. Implementing a CNN Using PyTorch

PyTorch provides several useful components:

| PyTorch component | Purpose |
|---|---|
| `nn.Conv2d()` | Performs two-dimensional convolution |
| `nn.ReLU()` | Applies the ReLU activation function |
| `nn.MaxPool2d()` | Performs max pooling |
| `nn.Flatten()` | Flattens feature maps |
| `nn.Linear()` | Defines fully connected layers |
| `nn.CrossEntropyLoss()` | Computes classification loss |

A simple CNN can be implemented by inheriting from `nn.Module` and defining its layers in the constructor.

## 5. Training a CNN

The training process follows the standard supervised learning procedure:

1. Load and preprocess the image dataset.
2. Define the CNN architecture.
3. Initialize the loss function and optimizer.
4. Perform a forward pass to obtain predictions.
5. Calculate the classification loss.
6. Compute gradients using backpropagation.
7. Update model parameters using the optimizer.
8. Evaluate the model on validation or test data.

The objective is to minimize the cross-entropy loss:

$$
\mathcal{L}=
-\frac{1}{N}
\sum_{i=1}^{N}
\log p_{i,y_i}
$$

where $p_{i,y_i}$ is the predicted probability assigned to the true class of sample $i$.

## 6. Applications

CNNs are widely used in:

- Image classification.
- Object detection.
- Image segmentation.
- Medical image analysis.
- Facial recognition.
- Satellite and remote sensing image analysis.

## 7. Conclusion

Convolutional Neural Networks are powerful architectures for learning spatial representations from image data. Their ability to extract local features through convolution, reduce dimensionality through pooling, and learn nonlinear relationships makes them particularly effective for computer vision tasks. PyTorch provides flexible tools for implementing, training, and evaluating CNN models.
