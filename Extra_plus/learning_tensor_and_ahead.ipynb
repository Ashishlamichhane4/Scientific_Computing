```python
import torch
import torch.nn as nn
```

```python
x = torch.randn(256, 4) 
print(x.shape,x.dtype)
```

```python
encoder = nn.Linear(4, 10)# input -> representation
print("This is encoder",encoder)


decoder = nn.Linear(10, 4) 
print("This is decoder",decoder)

```

```python
params = list(encoder.parameters()) + list(decoder.parameters())
opt = torch.optim.SGD(params, lr=0.01)

loss_fn = nn.MSELoss()
```

```python
for step in range(1001):
    z = torch.relu(encoder(x))           # the 10 output values
    x_hat = decoder(z)                   # try to rebuild the input
    loss = loss_fn(x_hat, x)             # how far off is the rebuild?

    opt.zero_grad()                      # clear old gradients
    loss.backward()                      # compute d(loss)/d(weights)
    opt.step()
    if step % 100 == 0:
      print(f"step {step}, loss {loss.item():.4f}")   # should go down 
```
