1.Function that generates the synthetic data following the perfect releationship
-------------------------------------------------------------------------------

Implementation details:
1. Use np.random.default rng(42) for reproducibility
2. Generate X values randomly between 0 and 1
3. Use the relationship: y= 2 + 3x (intercept=2, slope=3)
4.Return X as a 2D array with shape (n,1)

````python
import numpy as np

def create_perfect_linear_data(n_samples: int = 100):
    rng = np.random.default_rng(42)

    X = rng.random((n_samples, 1))
    y = 2 + 3 * X[:, 0]

    return X, y
````

42 is the seed. It makes the random numbers reproducible.
rng.random() generates random floating-point numbers between 0 and 1.
n_samples is a parameter that tells the function how many data points to generate.
(n_samples, 1) means n_samples rows and 1 column. For example, if n_samples = 100, then X.shape is (100, 1)

````python
X_linear, y_linear= create_perfect_linear_data()

print("X shape:", X_linear.shape)
print("y shape:", y_linear.shape)

print("First 5 X values:\n", X[:5])
print("First 5 y values:\n", y[:5])
````

````python


def create_noisy_linear_data(
    n_samples: int = 100,
    noise_std: float = 2
):
    rng = np.random.default_rng(42)

    X = rng.random((n_samples, 1))
    noise = rng.normal(0, noise_std, n_samples)

    y = 2 + 3 * X[:, 0] + noise

    return X, y
````

````python
def create_nonlinear_data(n_samples: int = 100):
    # Create a random number generator with seed 42 for reproducibility.
    rng = np.random.default_rng(42)

    # Generate n_samples random numbers between 0 (inclusive) and 1 (exclusive),
    # arranged in n_samples rows and 1 column.
    X = rng.random((n_samples, 1))

    # Calculate y by squaring each X value to create a perfect quadratic relationship.
    y = X[:, 0] ** 2

    # Return X as a 2D array and y as a 1D array.
    return X, y
````

### Analytical Solution for Learning Optimal Weights

````python


class LinearRegressionNormalEquation:

    def __init__(self):
        # Initialize weights as None because the model has not been trained yet.
        self.weights = None

    def fit(self, X: np.ndarray, y: np.ndarray):
        # Add a column of ones to X to include the intercept.
        X_augmented = np.column_stack([np.ones(len(X)), X])

        # Calculate X transpose multiplied by X.
        XT_X = X_augmented.T @ X_augmented

        # Calculate the inverse of X transpose multiplied by X.
        XT_X_inv = np.linalg.inv(XT_X)

        # Calculate the optimal weights using the normal equation.
        self.weights = XT_X_inv @ X_augmented.T @ y

    def predict(self, X: np.ndarray) -> np.ndarray:
        # Add a column of ones to the new input data.
        X_augmented = np.column_stack([np.ones(len(X)), X])

        # Multiply the augmented input data by the learned weights.
        y_pred = X_augmented @ self.weights

        return y_pred
````

### 3. Gradient Descent Implementation

Gradient descent iteratively minimizes the loss function:

$$
\boxed{
L(\mathbf{w}) = \frac{1}{2n}\sum_{i=1}^{n}(y_i-\mathbf{w}^T\mathbf{x}_i)^2
}
\tag{3}
$$

The gradient is:

$$
\boxed{
\nabla L(\mathbf{w}) = \frac{1}{n}X^T(X\mathbf{w}-\mathbf{y})
}
\tag{4}
$$

The update rule is:

$$
\boxed{
\mathbf{w}^{(t+1)} = \mathbf{w}^{(t)}-\alpha\nabla L(\mathbf{w}^{(t)})
}
\tag{5}
$$

Implement `LinearRegressionGradientDescent` using batch gradient descent.

**Implementation requirements:**

1. Augment the input matrix $X$ with a column of ones to include the intercept.
2. Initialize the weights to zero.
3. For each iteration:
   - Compute predictions: $\hat{\mathbf{y}} = X\mathbf{w}$
   - Compute the error: $\mathbf{e} = \hat{\mathbf{y}}-\mathbf{y}$
   - Compute the gradient: $\nabla L(\mathbf{w}) = \frac{1}{n}X^T\mathbf{e}$
   - Update the weights: $\mathbf{w} \leftarrow \mathbf{w}-\alpha\nabla L(\mathbf{w})$
   - Calculate and store the loss: $L = \frac{1}{2}\operatorname{mean}(\mathbf{e}^2)$
4. Implement the `predict` method, following the same approach as `LinearRegressionNormalEquation`.

**Constructor parameters:**
- `learning_rate: float = 0.01` — Step size $\alpha$.
- `n_iterations: int = 1000` — Number of optimization steps.

**Attributes:**
- `learning_rate: float`
- `n_iterations: int`
- `weights: ndarray | None`
- `loss_history: list[float]` — Tracks the loss over iterations.

### 3.1 Gradient Descent Pseudocode

**Pseudocode: Batch Gradient Descent**

**Input:**
- Training data: `X`
- Target values: `y`
- Learning rate: `learning_rate = 0.01`
- Number of iterations: `n_iterations = 1000`

**Output:**
- Learned weights: `weights`
- Loss history: `loss_history`

**Algorithm:**

1. **Initialize the model**
   - Add a column of ones to `X` to include the intercept.
   - Initialize all weights to zero.
   - Initialize `loss_history` as an empty list.

2. **For each iteration from `1` to `n_iterations`:**
   - Calculate predictions:
     ```text
     y_pred = X_augmented @ weights
     ```
   - Calculate the prediction error:
     ```text
     error = y_pred - y
     ```
   - Calculate the gradient:
     ```text
     gradient = (X_augmented.T @ error) / n
     ```
   - Update the weights:
     ```text
     weights = weights - learning_rate * gradient
     ```
   - Calculate the loss:
     ```text
     loss = 0.5 * mean(error**2)
     ```
   - Append the loss to `loss_history`.

3. **Return** the learned weights and loss history.

**Prediction Algorithm:**

1. Add a column of ones to the new input data `X`.
2. Calculate predictions:
   ```text
   y_pred = X_augmented @ weights
   ```
3. Return `y_pred`.

````python

class LinearRegressionGradientDescent:

    def __init__(self, learning_rate: float = 0.01, n_iterations: int = 1000):
        self.learning_rate = learning_rate
        self.n_iterations = n_iterations
        self.weights = None
        self.loss_history = []

    def fit(self, X: np.ndarray, y: np.ndarray):

        # a) Add a column of ones for the intercept
        X_augmented = np.column_stack([np.ones(len(X)), X])

        # b) Initialize weights to zero
        self.weights = np.zeros(X_augmented.shape[1])

        # c) Perform batch gradient descent
        for _ in range(self.n_iterations):

            # Calculate predictions
            y_pred = X_augmented @ self.weights

            # Calculate errors
            error = y_pred - y

            # Calculate gradient
            gradient = (X_augmented.T @ error) / len(y)

            # Update weights
            self.weights = self.weights - self.learning_rate * gradient

            # Calculate and store loss
            loss = 0.5 * np.mean(error**2)
            self.loss_history.append(loss)

    def predict(self, X: np.ndarray) -> np.ndarray:

        # Add a column of ones
        X_augmented = np.column_stack([np.ones(len(X)), X])

        # Calculate predictions using learned weights
        return X_augmented @ self.weights
````

````python



class LinearRegressionSGD:
    """Linear regression trained with mini-batch stochastic gradient descent.

    Per batch (X_b, y_b) with m samples:
        e  = X_b w - y_b
        dL = 1/m * X_b^T e
        w := w - alpha * dL

    The intercept is handled by augmenting X with a column of ones as the
    FIRST column, so weights[0] is the intercept.
    """

    def __init__(
        self,
        learning_rate: float = 0.01,
        n_iterations: int = 1000,
        batch_size: int = 32,
    ):
        self.learning_rate: float = learning_rate
        self.n_iterations: int = n_iterations  # number of epochs
        self.batch_size: int = batch_size
        self.weights: np.ndarray | None = None
        self.loss_history: list[float] = []

    def fit(self, X: np.ndarray, y: np.ndarray) -> "LinearRegressionSGD":
        y = np.asarray(y, dtype=float).ravel()
        self.loss_history = []

        # a) augment X with intercept column
        X_augmented = np.column_stack([np.ones(len(X)), X])
        n_samples, n_params = X_augmented.shape

        # b) initialize weights to zeros
        self.weights = np.zeros(n_params)

        # c) loop over epochs
        for _ in range(self.n_iterations):
            # shuffle the data
            indices = np.random.permutation(n_samples)
            X_shuffled = X_augmented[indices]
            y_shuffled = y[indices]

            epoch_loss = 0.0

            # loop over mini-batches
            for start_idx in range(0, n_samples, self.batch_size):
                end_idx = min(start_idx + self.batch_size, n_samples)
                X_batch = X_shuffled[start_idx:end_idx]
                y_batch = y_shuffled[start_idx:end_idx]

                # batch predictions and error
                error = X_batch @ self.weights - y_batch

                # batch gradient (average over the samples in this batch)
                gradient = (X_batch.T @ error) / len(y_batch)

                # update weights
                self.weights = self.weights - self.learning_rate * gradient

                # accumulate squared error (sum, so the last smaller batch
                # is weighted correctly)
                epoch_loss += 0.5 * np.sum(error ** 2)

            # average loss over all samples seen in this epoch
            self.loss_history.append(epoch_loss / n_samples)

        return self

    def predict(self, X: np.ndarray) -> np.ndarray:
        if self.weights is None:
            raise ValueError("Model is not fitted yet. Call fit() first.")
        X_augmented = np.column_stack([np.ones(len(X)), X])
        return X_augmented @ self.weights


if __name__ == "__main__":
    np.random.seed(0)
    X = np.random.randn(200, 3)
    true_w = np.array([4.0, 2.0, -1.0, 0.5])  # intercept first
    y = true_w[0] + X @ true_w[1:] + 0.1 * np.random.randn(200)

    model = LinearRegressionSGD(learning_rate=0.05, n_iterations=100, batch_size=32)
    model.fit(X, y)

    X_aug = np.column_stack([np.ones(len(X)), X])
    w_exact = np.linalg.solve(X_aug.T @ X_aug, X_aug.T @ y)

    print("SGD weights:   ", np.round(model.weights, 4))
    print("Exact weights: ", np.round(w_exact, 4))
    print("Max difference:", np.max(np.abs(model.weights - w_exact)))
    print("Epochs recorded:", len(model.loss_history))
    print("Loss first/last:", model.loss_history[0], model.loss_history[-1])

    # edge cases: batch larger than dataset, batch size that doesn't divide n
    big = LinearRegressionSGD(learning_rate=0.1, n_iterations=300, batch_size=1000).fit(X, y)
    odd = LinearRegressionSGD(learning_rate=0.05, n_iterations=100, batch_size=33).fit(X, y)
    print("batch>n  max diff:", np.max(np.abs(big.weights - w_exact)))
    print("batch=33 max diff:", np.max(np.abs(odd.weights - w_exact)))
    print("predict shape:", model.predict(X[:5]).shape)
````

### R² Score Metric

````python


def calculate_r2(y_actual, y_predicted):

    # Calculate the mean of actual values
    y_mean = np.mean(y_actual)

    # Calculate Sum of Squared Errors (SSE)
    sse = np.sum((y_actual - y_predicted) ** 2)

    # Calculate Total Sum of Squares (SST)
    sst = np.sum((y_actual - y_mean) ** 2)

    # Calculate R² score
    r2 = 1 - (sse / sst)

    return r2
````

````python
y_actual = np.array([100, 140, 180, 220])
y_predicted = np.array([95, 125, 155, 185])

r2 = calculate_r2(y_actual, y_predicted)

print("R² Score:", r2)
````

### RMSE Metric

````python


def compute_rmse(y_true: np.ndarray, y_pred: np.ndarray) -> float:

    # Calculate Mean Squared Error
    mse = np.mean((y_true - y_pred) ** 2)

    # Return Root Mean Squared Error
    return np.sqrt(mse)
````

````python
y_true = np.array([100, 140, 180, 220])
y_pred = np.array([95, 125, 155, 185])

rmse = compute_rmse(y_true, y_pred)

print("RMSE:", rmse)
````
