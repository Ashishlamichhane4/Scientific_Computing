```python
#Categorical data in each of the columns
for col in initial_df.select_dtypes(include=["object", "category"]).columns:
    print(f"{col}: {initial_df[col].unique()}")
```

```python
# Categorical Encoding
binary_cols = [
    "mainroad",
    "guestroom",
    "basement",
    "hotwaterheating",
    "airconditioning",
    "prefarea"
]

initial_df[binary_cols] = (
    initial_df[binary_cols]
    .replace({"yes": 1, "no": 0})
    .astype("int64")
)
```

```python
initial_df.head()
```

```python

initial_df["furnishingstatus"] = initial_df["furnishingstatus"].map({
    "unfurnished": 0,
    "semi-furnished": 1,
    "furnished": 2
})
```

```python
initial_df
```

```python
#separating features and the targe
X = initial_df.drop(columns=["price"])
y = initial_df["price"]
```

```python
total_samples = len(X)

train_size = int(0.8 * total_samples)
test_size = total_samples - train_size

print("Total samples:", total_samples)
print("Training samples:", train_size)
print("Testing samples:", test_size)
```

```python


np.random.seed(42)

indices = np.random.permutation(len(X))

train_indices = indices[:train_size]
test_indices = indices[train_size:]
```

```python
X_train = X.iloc[train_indices]
X_test = X.iloc[test_indices]

y_train = y.iloc[train_indices]
y_test = y.iloc[test_indices]

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

### Standardrization

```python
numeric_cols = [
    "area",
    "bedrooms",
    "bathrooms",
    "stories",
    "parking"
]
```

```python
means = X_train[numeric_cols].mean()
stds = X_train[numeric_cols].std()
```

```python
X_train[numeric_cols] = (
    X_train[numeric_cols] - means
) / stds

X_test[numeric_cols] = (
    X_test[numeric_cols] - means
) / stds
```
